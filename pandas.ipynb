```python
import seaborn as sns
lab = sns.load_dataset('titanic')
print(lab.columns)
print(lab.shape)
print(lab.dtypes)
```

```python

# 1. Missing values in 'deck'
missing_deck = lab['deck'].isna().sum()
print('Missing in deck:', missing_deck)

# 2. Drop 'deck' column
lab = lab.drop(columns=['deck'])

# 3. Fill missing 'embark_town' with the mode
lab['embark_town'] = lab['embark_town'].fillna(lab['embark_town'].mode()[0])

# 4. Confirm
missing_embark_town = lab['embark_town'].isna().sum()
print('Missing in embark_town after fillna:', missing_embark_town)

```

```python
# 1. Filter survivors who are female
survivors_f = lab[(lab['survived'] == 1) & (lab['sex'] == 'female')]

# 2. Sort by age ascending, print youngest 5
print(survivors_f.sort_values(by='age', ascending=True).head())

# 3. Count passengers per embark_town
print('Embark town counts:')
print(lab['embark_town'].value_counts())
```

```python
# 1. age_group via apply
def label_age(age):
    if age < 18:
        return 'child'
    elif age < 60:
        return 'adult'
    else:
        return 'senior'

lab['age_group'] = lab['age'].apply(label_age)

# 2. value_counts of age_group
print('Age group counts:')
print(lab['age_group'].value_counts())

# 3. family_size and survival rate by family_size
lab['family_size'] = lab['sibsp'] + lab['parch'] + 1
print('Survival rate by family size:')
print(lab.groupby('family_size')['survived'].mean())
```

```python
import pandas as pd

# 1. Fare pivot (pclass x embark_town)
fare_pivot = lab.pivot_table(index='pclass', columns='embark_town', values='fare', aggfunc='mean')
print('Fare pivot (pclass x embark_town):')
print(fare_pivot)

# 2. Build df_a, df_b, then merge on 'class' with how='inner'
df_a = pd.DataFrame({
    'class': ['First', 'Second', 'Third'],
    'deck_level': ['Upper', 'Middle', 'Lower']
})

df_b = pd.DataFrame({
    'class': ['First', 'Second', 'Third'],
    'avg_fare': [84.15, 20.66, 13.68]
})

merged = df_a.merge(df_b, on='class', how='inner')
print(merged)
```

```python
import pandas as pd

# 1. Uppercase 'who' column
lab['who_upper'] = lab['who'].str.upper()

# 2. Check if 'class' contains 'i' (case-insensitive)
lab['has_i'] = lab['class'].str.contains('i', case=False)

# 3. Day names from custom dates
dates = pd.Series(['2024-01-15', '2024-07-04', '2024-12-25'])
dates = pd.to_datetime(dates)
print('Day names:')
print(dates.dt.day_name())
```
